```python
!pip install -q tensorflow
import tensorflow as tf

# Create tensors
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

print("Tensor A:", a)
print("Tensor B:", b)

# Basic operations
print("Addition:", tf.add(a, b))
print("Subtraction:", tf.subtract(a, b))
print("Multiplication:", tf.multiply(a, b))

# Reshape tensor
print("Reshape:", tf.reshape(a, [3, 1]))

# Eager execution
print("Eager:", tf.executing_eagerly())

# Graph execution
@tf.function
def add(x, y):
    return x + y

print("Graph:", add(a, b))
```

```python
!pip install -q pandas scikit-learn
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    'Age': [20, 22, None, 24],
    'Marks': [80, 90, 70, 100],
    'City': ['Pune', 'Mumbai', 'Pune', 'Nashik']
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Fill missing Age with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Normalize Age using MinMaxScaler
scaler = MinMaxScaler()
df[['Age']] = scaler.fit_transform(df[['Age']])

# Convert City into numbers
df['City'] = df['City'].map({'Pune': 0, 'Mumbai': 1, 'Nashik': 2})

print("\nCleaned Data:")
print(df)
```

```python
!pip install -q pandas matplotlib seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Age': [18, 20, 22, 24, 26],
    'Marks': [60, 70, 80, 90, 95],
    'Hours': [2, 3, 4, 5, 6]
}

df = pd.DataFrame(data)

# Histogram
plt.hist(df['Marks'])
plt.title("Histogram")
plt.show()

# Scatter plot
plt.scatter(df['Age'], df['Marks'])
plt.title("Scatter Plot")
plt.show()

# Heatmap
sns.heatmap(df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
!pip install -q gensim
from gensim.models import Word2Vec

data = [
    ["king", "queen", "prince", "royal"],
    ["cat", "dog", "animal", "pet"],
    ["computer", "software", "technology"],
    ["python", "java", "programming", "language"]
]

# Train Word2Vec
model = Word2Vec(data, vector_size=10, min_count=1, epochs=200)

# Find similar words
print("Similar words:", model.wv.most_similar("king", topn=2))

# Analyze semantic relationships
print("King and Queen similarity:",
      model.wv.similarity("king", "queen"))

print("King and Computer similarity:",
      model.wv.similarity("king", "computer"))
```

```python

!pip install -q transformers peft
!pip install -q --upgrade torchao peft
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import LoraConfig, get_peft_model

# Load model
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilgpt2")
model = AutoModelForCausalLM.from_pretrained("distilbert/distilgpt2")

tokenizer.pad_token = tokenizer.eos_token

# Apply LoRA
config = LoraConfig(r=4, lora_alpha=8, target_modules=["c_attn"])
model = get_peft_model(model, config)

# Generate text
input_text = tokenizer("Artificial intelligence is", return_tensors="pt")
output = model.generate(**input_text, max_new_tokens=30)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
!pip install -q transformers rouge-score

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from rouge_score import rouge_scorer

# Load model
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

# Input text
text = "AI helps computers learn and solve problems."
prompt = "Summarize: " + text

# Generate summary
inputs = tokenizer(prompt, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=30)
result = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("Summary:", result)

# Evaluate using ROUGE
scorer = rouge_scorer.RougeScorer(["rouge1"])
print("ROUGE Score:", scorer.score(text, result))
```

```python

!pip install faiss-cpu sentence-transformers transformers

import faiss
from sentence_transformers import SentenceTransformer
from transformers import pipeline

docs = ["Python is a language", "AI means Artificial Intelligence"]

m = SentenceTransformer('all-MiniLM-L6-v2')
v = m.encode(docs)

db = faiss.IndexFlatL2(384)
db.add(v)

q = m.encode(["What is AI?"])
_, i = db.search(q, 1)

# Using 'distilgpt2' which is compatible with the text-generation pipeline
llm = pipeline("text-generation", model="distilgpt2")
print(llm(docs[i[0][0]], max_new_tokens=20))
```

```python
!pip install -q transformers torch
from transformers import AutoModel
import torch

m = AutoModel.from_pretrained("distilbert/distilbert-base-uncased")

q = torch.quantization.quantize_dynamic(
    m, {torch.nn.Linear}, dtype=torch.qint8)

print("Original Size:", sum(p.numel() for p in m.parameters()))
print("Quantized Model:", q)
```
